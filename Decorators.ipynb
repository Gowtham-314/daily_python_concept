```python
import time
```

```python
def timer_dec(base_fn):
    def enhanced_fn():
        start_time = time.time()
        base_fn()
        end_time = time.time()
        print(f"Time taken: {end_time - start_time:.2f} seconds")
    return enhanced_fn

@timer_dec
def brew_tea():
    print("Boiling water...")
    time.sleep(1)  # Simulating time taken to boil water
    print("Tea is ready!")
    
    
@timer_dec
def brew_coffee():
    print("Grinding coffee beans...")
    time.sleep(2)  # Simulating time taken to grind coffee
    print("Coffee is ready!")
```

```python
timer_dec(brew_tea)  #Manually decorating the function
timer_dec(brew_coffee)  #Manually decorating the function
```

```python
decorated_brew_tea = timer_dec(brew_tea)
decorated_brew_tea()        # Calling the decorated function
```
