```python
import time
```

```python
def timer_dec(base_fn):
    def enhanced_fn():
        start_time = time.time()
        base_fn()
        end_time = time.time()
        print(f"Time taken: {end_time - start_time:.2f} seconds")
    return enhanced_fn

@timer_dec
def brew_tea():
    print("Boiling water...")
    time.sleep(1)  # Simulating time taken to boil water
    print("Tea is ready!")

brew_tea()
```
