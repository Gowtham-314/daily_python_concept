#### ```9. Palindrome Number```

````python
class Solution(object):
    def isPalindrome(self, x):
        """
        :type x: int
        :rtype: bool
        """
        
        rev = int(str(abs(x))[::-1])
        
        if rev == x:
            return True
        else:
            return False
            
````

````python
s = Solution()

s.isPalindrome(-121)
        
````

````python
class Solution(object):
    def isPalindrome(self, x):
        """
        :type x: int
        :rtype: bool
        """
        
        num = abs(x)
        rev = 0
        
        while num > 0:
            last = num % 10
            rev = ( rev * 10 ) +last
            num = num // 10
            
                    
        
        return True if rev == x else False 
            

s = Solution()

s.isPalindrome(-121)
        
````
