```pathlib```
###### pathlib is a built-in Python module used to work with files and folders (paths) in an easy, object-oriented way.

````python
from pathlib import Path
````

````python
path=Path("sample.txt")
print(path)
````

````python
pdf_path = Path("pdfs") / "questions.pdf"
print(pdf_path)
````

```Check if File Exists```

````python
file = Path("sample.pdf")
print(file.exists())
````

```Check if it is a File```

````python
file = Path("sample.pdf")
print(file.is_file())
````

```Check if it is a Folder```

````python
folder = Path("data")
print(folder.is_dir())
````

```Get File Name```

````python
file = Path("data/python_notes.pdf")
print(file.name)
````

```Get File Name Without Extension```

````python
file = Path("data/python_notes.pdf")
print(file.stem)
````

```Get File Extension```

````python
file = Path("data/python_notes.pdf")
print(file.suffix)
````

```Create a Folder```

````python
folder = Path("uploads")
folder.mkdir(exist_ok=True)

#exist_ok=True prevents an error if the folder already exists.
````

```Read a Text File```

````python
file = Path("sample.txt")
content = file.read_text()
print(content)
````

```Write a Text File```

````python
file = Path("notes.txt")
file.write_text("Hello Python")
````

```Get All PDF Files in a Folder```

````python
folder = Path("file_path")

for pdf in folder.glob("*.pdf"):
    print(pdf)
````

```Search Inside All Subfolders```

````python
for pdf in Path("pdfs").rglob("*.pdf"):
    print(pdf)
````

```Search the entire system```

````python
filename = "questions.pdf"
for file in Path.home().rglob(filename):
    if file.exists():
        print(f"Found: {file}")
        break
````
