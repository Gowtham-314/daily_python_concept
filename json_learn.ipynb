## JSON Learning

```python
import json
```

```python
from pprint import pprint

with open("data.json", "r") as file:
    data = json.load(file)
    
    
    pprint(data["products"],width=40)
```

```python
print(data["products"][0]["name"])
```

```python
with open("data.json", "r") as file:
    data = json.load(file)
    
for product in data["products"]:
    print(product["sale_price"])
```

```python
with open("data.json", "r") as file:
    data = json.load(file)
    
for product in data["products"]:
    print(product.get("sale_price"))
```

```python
with open("data.json", "r") as file:
    data = json.load(file)
    data["products"][0].get("sizes").append("extra large")

with open("data.json", "w") as file:
    json.dump(data, file, indent=2)

```

```python
with open("student_data.json", "r") as file:
    data = json.load(file)
```

```python
for student in data["students"]:
    print(student["name"])
```

```python
for student in data["students"]:
    print(student["usn"])
```

```python
for student in data["students"]:
    print(f"{student['name']} : {student['usn']}")
```

```python
for student in data["students"]:
    if student["mentor"] == "Dr. Riyaz UR Rehman A":
        print(f"{student['name']} : {student['email']}")
```

```python
for student in data["students"]:
    if "gowda" in student["name"].lower():
        print(student["name"])
```

```python
for student in data["students"]:
    if "p" in student["name"].lower():
        if student["name"].lower().startswith("p"):
            print(student["name"])
```
