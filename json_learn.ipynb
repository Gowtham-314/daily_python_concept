## JSON Learning

```python
import json
```

```python
from pprint import pprint

with open("data.json", "r") as file:
    data = json.load(file)
    
    
    pprint(data["products"],width=40)
```

```python
print(data["products"][0]["name"])
```

```python
with open("data.json", "r") as file:
    data = json.load(file)
    
for product in data["products"]:
    print(product["sale_price"])
```

```python
with open("data.json", "r") as file:
    data = json.load(file)
    
for product in data["products"]:
    print(product.get("sale_price"))
```

```python
with open("data.json", "r") as file:
    data = json.load(file)
    data["products"][0].get("sizes").append("extra large")

with open("data.json", "w") as file:
    json.dump(data, file, indent=2)

```
