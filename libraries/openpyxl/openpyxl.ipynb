```python
from openpyxl import Workbook

# Create a workbook
wb = Workbook()

# Select the active worksheet
ws = wb.active
ws.title = "Students"

# Add data
ws["A1"] = "Name"
ws["B1"] = "Age"

ws["A2"] = "Alice"
ws["B2"] = 20

ws["A3"] = "Bob"
ws["B3"] = 22

# Save the workbook
wb.save("students.xlsx")
```
