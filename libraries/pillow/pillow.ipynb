```python
from PIL import Image
from pathlib import Path

file = Path.cwd().parent / "qrcode"
```

```python
# Open the image file
File = Image.open(file/"email.png")
File.show()
```

```python
img = Image.open(f"{file}/email.png")
img.show()
```

```python
# Image info
print(img.format)
print(img.size)
print(img.mode)
```

```python
# Create a Image

img = Image.new("RGB", (500, 500), "blue")
img.show()
img.save("output.png")
```

```python
# Resize the image
resized = img.resize((300, 300))
resized.show()
```

```python
# Crop Image
cropped = img.crop((100, 100, 400, 400))
cropped.show()
```
