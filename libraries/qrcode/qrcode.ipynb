````python
import qrcode

qr = qrcode.QRCode(
    version=1,
    error_correction=qrcode.constants.ERROR_CORRECT_H,
    box_size=10,
    border=4
)

qr.add_data("https://github.com/Gowtham-314")

qr.make(fit=True)

img = qr.make_image(
    fill_color="#333333",
    back_color="#86cba8",
    edge_color="#ff5733"
)

img.save("github.png")
````
