```Matplotlib```

````python
import matplotlib.pyplot as plt
````

````python
a = [2023,2024,2025,2026]
b = [50,75,100,125]
````

````python
plt.plot(a,b,marker='o',color='green',linestyle='--')
plt.xlabel('Year')     
plt.ylabel('Value')
plt.title('Line Plot Example')
plt.grid(True)
plt.show()
````

````python
import numpy as np
````

````python
x = np.array([2023,2024,2025,2026])
y = np.array([50,75,100,15])
plt.plot(x,y,marker='o',
        color='green',
        linestyle='dashdpt',
        markersize=10,
        markerfacecolor='red',
        markeredgecolor='blue'   ''
         )
# markersize : ms
# markerfacecolor : mfc
# markeredgecolor : mec
plt.xlabel('Year')
plt.ylabel('Value')
plt.title('Line Plot Example')
plt.grid(True)
plt.show()
````
