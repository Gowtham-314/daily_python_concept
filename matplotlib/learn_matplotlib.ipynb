````python
import matplotlib.pyplot as plt
````

````python
a = [2023,2024,2025,2026]
b = [50,75,100,125]
````

````python
plt.plot(a,b,marker='o',color='green',linestyle='--')
plt.xlabel('Year')     
plt.ylabel('Value')
plt.title('Line Plot Example')
plt.grid(True)
plt.show()
````

````python
import numpy as np
````

````python
x = np.array([2023,2024,2025,2026])
y = np.array([50,75,100,15])
plt.plot(x,y,marker='o',
        color='green',
        linestyle='dashed',  # 'dashed' or '--' , 'dotted' or ':','dashdot' or '-.','solid'
        markersize=20,
        markerfacecolor='red',
        markeredgecolor='blue',
        linewidth=3
         )


# markersize : ms
# markerfacecolor : mfc
# markeredgecolor : mec
plt.xlabel('Year')
plt.ylabel('Value')
plt.title('Line Plot Example')
plt.grid(True)
plt.show()
````

````python
x = np.array([2023,2024,2025,2026])
y = np.array([50,75,100,15])
y1 = np.array([60,99,52,130])
y2 = np.array([30,55,65,90])


line_styles = dict(marker='o',
                   linestyle='dashed',  
                   markersize=20,
                   markerfacecolor='red',
                   markeredgecolor='blue',
                   linewidth=3
                   )

plt.plot(x,y,**line_styles,color='green')

plt.plot(x,y1,**line_styles,color='blue')

plt.plot(x,y2,**line_styles,color='orange')


# markersize : ms
# markerfacecolor : mfc
# markeredgecolor : mec
plt.xlabel('Year')
plt.ylabel('Value')
plt.title('Line Plot Example')
plt.grid(True)
plt.show()
````

``` plt.xticks(x)  # Set x-axis ticks to the values in x ```

````python
# labels

x = np.array([2023,2024,2025,2026])
y = np.array([50,75,100,15])
y1 = np.array([60,99,52,130])
y2 = np.array([30,55,65,90])


line_styles = dict(marker='o',
                   linestyle='dashed',  
                   markersize=20,
                   markerfacecolor='red',
                   markeredgecolor='blue',
                   linewidth=3
                   )

font_style = dict(fontstyle='italic',
                  fontweight='bold',
                  family='serif',
                  fontsize=14
                    )

plt.plot(x,y,**line_styles,color='green')

plt.plot(x,y1,**line_styles,color='blue')

plt.plot(x,y2,**line_styles,color='orange')

plt.xticks(x)  # Set x-axis ticks to the values in x

plt.tick_params(axis="both",  # axis='x', axis='y'
                colors="blue")  # Color to ticks



plt.xlabel('Year',color='#2e9e80', **font_style)
plt.ylabel('Value',color='#85fa00', **font_style)
plt.title('Line Plot Example',color='purple', **font_style)
plt.show()
````

````python

plt.plot(x,y)
plt.grid(axis="y",
         linewidth=2,
         color="lightgreen",
         linestyle="--")
plt.show()
````
