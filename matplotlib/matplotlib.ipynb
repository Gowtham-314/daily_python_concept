```Matplotlib```

````python
import matplotlib.pyplot as plt
````

````python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [100, 150, 200, 250, 300, 350]
````

```Line plot```

````python
plt.plot(months, sales)
plt.show()


# plot() draws a line connecting the data points
````

````python
#  Customizing the line plot

plt.plot(months, sales, marker='o', linestyle='--', color='green')
plt.show()
````

````python
# Labels , title and Legend

plt.plot(months, sales, marker='o', linestyle='--', color='green', label='Sales')
plt.xlabel('Months')
plt.ylabel('Sales')
plt.title('Sales Data')
plt.legend()
plt.show()

# Labels and titles make charts easier to understand.
````

````python
# Bar Chart

plt.bar(months, sales, color='lightgreen')
plt.show()

# bar() shows values as rectangular bars.
````

````python
# Histogram

marks = [85, 90, 78, 92, 88, 95, 80, 82, 91, 89]
plt.hist(marks, bins=5, color='lightblue', edgecolor='black')   
plt.show()
````

````python
# Scatter plot

age = [21,22,20,23,19,22]
marks=[88,90,78,96,81,67]

plt.scatter(age,marks,color="red")
plt.show()

# scatter() shows the relationship between two variables.
````

````python
# Pie Chart

sizes = [40,30,20,10]
labels=["Python","C","Java","C++"]
plt.pie(sizes,labels=labels,autopct="%1.1f%%")
plt.show()
````

````python
# Multiple Lines

sales_2023 = [250,320,280,420,360,390]
sales_2024 = [300,350,310,500,420,450]

plt.plot(months,sales_2023, marker='o', linestyle='--', color='green', label='2023 Sales')
plt.plot(months,sales_2024, marker='s', linestyle='-', color='blue', label='2024 Sales')
plt.legend()
plt.show()
````

````python
# Subplots

plt.subplot(1,2,1)  # 1 rows, 2 column, first plot
plt.plot(months, sales, marker='o', linestyle='--', color='green')
plt.subplot(1,2,2)  # 1 row, 2 columns, second plot
plt.bar(months, sales, color='lightgreen')
plt.show()
````

````python
# Grid and Style
plt.style.use('ggplot')
plt.plot(months, sales_2023, marker='o', linestyle='--', color='green')
plt.grid(True)
plt.show()
````

````python
# Figure Size

plt.figure(figsize=(8,4))
plt.plot(months, sales_2023, marker='o', linestyle='--', color='green')
plt.show()
````
