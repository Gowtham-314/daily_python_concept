### ```Pandas with Matplotlib```

````python
from matplotlib import pyplot as plt
import numpy as np
import pandas as pd
````

````python
df = pd.read_csv('bank.csv')
````

````python
df["marital"].value_counts()
````

````python
type_count = df["marital"].value_counts()

colors = ["#7A54C5", "#4ED4AE", "#AFD448"]

plt.bar(type_count.index, type_count.values, color=colors,
        edgecolor='black'
        )
plt.xlabel("Marital Status")
plt.ylabel("Count")
plt.title("Distribution of Marital Status")
plt.show()
````

````python
type_count = df["job"].value_counts()

colors = ["#7A54C5", "#4ED4AE", "#AFD448", "#F5A623","#FF6B6B", "#4A90E2", "#50E3C2", "#F8E71C",]

plt.barh(type_count.index, type_count.values, color=colors,
        edgecolor='black'
        )
plt.xlabel("Job Count")
plt.ylabel("Job Type")
plt.title("Distribution of Job Types")
plt.show()
````

````python
type_count = df["education"].value_counts()

colors = ["#6DE27E", "#4E8FD4", "#7D66DA", "#F5A623"]

plt.pie(type_count.values, labels=type_count.index,
        colors=colors, 
        autopct='%1.1f%%',
        startangle=90,
        shadow=True,
        wedgeprops={'edgecolor': 'black'}
        )
plt.title("Distribution of Education Levels")
plt.legend()
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
````

#### ```Save the Plot as a Image```

````python
type_count = df["education"].value_counts()

colors = ["#6DE27E", "#4E8FD4", "#7D66DA", "#F5A623"]

plt.pie(type_count.values, labels=type_count.index,
        colors=colors, 
        autopct='%1.1f%%',
        startangle=90,
        shadow=True,
        wedgeprops={'edgecolor': 'black'}
        )
plt.title("Distribution of Education Levels")
plt.legend()
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.savefig("education_distribution.png")  # Save the figure as a PNG file
````
