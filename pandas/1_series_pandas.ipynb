````python
import pandas as pd
````

```series```

````python
data1 = [100.4, 102, 104, 106, 108]
series1 = pd.Series(data1, index=['B1', 'B2', 'B3', 'B4', 'B5'])
print(series1)
````

````python
data = [105, 109, 104, 110, 112]
series = pd.Series(data, index=['A1', 'A2', 'A3', 'A4', 'A5'])
print(series)
````

````python
series.loc["A3"] = 320
````

````python
print(series.loc["A3"])  # Accessing the value at index 'A3'
````

````python
print(series.iloc[0])  # Accessing the value at position 1
````

````python
print(series[series > 200])  # Filtering values greater than 200
````

````python
data_daily = {
    "Day 1": 1750,
    "Day 2": 2100,
    "Day 3": 1700,
}

series = pd.Series(data_daily)

series.loc["Day 1"] = 1800 

print('Day 1   ', series.loc["Day 1"])

print(series[series > 1800])  # Filtering values greater than 1800
````

````python
series.loc["Day 4"] = 1900  # Adding a new entry for Day 4
print(series)
````

````python
print(series.sort_values())  # Sorting the series by values
````
