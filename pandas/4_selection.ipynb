```python
# Selections in Pandas

import pandas as pd
```

```python
df = pd.read_json('data_sets/students.json')
```

```python
# Selection by column name
print(df['auid'].to_string())
```

```python
# print(pd['usn'].to_string())

print(df[['auid' , 'name' , 'mentor']].to_string())
```

```python
# Selection by Row /s

print(df.loc[1].to_string())
```

```python
df = pd.read_csv('data_sets/data.csv', index_col='Name')
print(df)
```

```python
print(df.loc["Pikachu", ["Type1","Height","Weight"]].to_string())
```

```python
print(df.loc["Charizard":"Blastoise", ["Type1","Height","Weight"]].to_string())
```

```python
print(df.iloc[2])   # Selection by position
```

```python
print(df.iloc[0:11])   # Selection by position with slicing
```

```python
print(df.iloc[0:11:2])   # Selection by position with slicing and step
```

```python
print(df.iloc[0:11:2, 0:3])
```

```python
pokemon = input("Enter the name of the pokemon: ")

try :
    print(df.loc[pokemon, ["Type1","Height","Weight"]])
except KeyError:
    print(f"{pokemon} not found in the dataset.")
```
