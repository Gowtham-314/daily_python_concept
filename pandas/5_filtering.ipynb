```python
# Filtering = Keeping the rows that match a condition.
import pandas as pd
```

```python
df = pd.read_csv('data_sets/bank.csv')
```

```python
less_balance = df[df["balance"] < 0]
print(less_balance.to_string())
```

```python
heavy_balance = df[df["balance"] == max(df["balance"])]
print(heavy_balance.to_string())
```

```python
low_balance = df[df["balance"] == min(df["balance"])]
print(low_balance.to_string())
```

```python
print(df[df["age"] == 30])
```

```python
loan_df=df[df["education"] == "unknown"]
print(loan_df)
```

```python
loan_df=df[df["education"] == "unknown"]
print(loan_df[["age","job","marital","education"]])
```

```python
loan_df=df[(df["education"] == "unknown") & (df["marital"] == "single")]
print(loan_df)
```

```python
loan_df=df[(df["education"] == "unknown") & (df["marital"] == "married") & (df["job"] == "blue-collar")]
print(loan_df)
```

```python
loan_df=df[(df["education"] == "unknown") & (df["marital"] == "married") & (df["job"] == "blue-collar")]

list = loan_df[["age","job","marital","education"]]

print(list.sort_values(by="age"))
```

```python
import matplotlib.pyplot as plt
```

```python
plt.pie(df["marital"].value_counts(), labels=df["marital"].value_counts().index, 
        autopct="%1.1f%%",
        shadow=True,
        colors=["#ff9999","#66b3ff","#99ff99"],
        startangle=90,
        explode=[0,0, 0.1],
        wedgeprops={'edgecolor': 'black'}
        )
plt.suptitle("Bank Customers Marital Status", fontsize=16, fontweight="bold")
plt.title("Distribution of Marital Status",family="Times New Roman", fontsize=14)
plt.legend(title="Marital Status", loc="upper right")
plt.axis("equal")
plt.show()
```
