```python
# Filtering = Keeping the rows that match a condition.
import pandas as pd
```

```python
df = pd.read_csv('data_sets/bank.csv')
```

```python
less_balance = df[df["balance"] < 0]
print(less_balance.to_string())
```

```python
heavy_balance = df[df["balance"] == max(df["balance"])]
print(heavy_balance.to_string())
```
