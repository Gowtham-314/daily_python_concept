```python
# Filtering = Keeping the rows that match a condition.
import pandas as pd
```

```python
df = pd.read_csv('data_sets/bank.csv')
```

```python
less_balance = df[df["balance"] < 0]
print(less_balance.to_string())
```

```python
heavy_balance = df[df["balance"] == max(df["balance"])]
print(heavy_balance.to_string())
```

```python
low_balance = df[df["balance"] == min(df["balance"])]
print(low_balance.to_string())
```

```python
print(df[df["age"] == 30])
```

```python
loan_df=df[df["education"] == "unknown"]
print(loan_df)
```

```python
loan_df=df[df["education"] == "unknown"]
print(loan_df[["age","job","marital","education"]])
```

```python
loan_df=df[(df["education"] == "unknown") & (df["marital"] == "single")]
print(loan_df)
```

```python
loan_df=df[(df["education"] == "unknown") & (df["marital"] == "married") & (df["job"] == "blue-collar")]
print(loan_df)
```

```python
loan_df=df[(df["education"] == "unknown") & (df["marital"] == "married") & (df["job"] == "blue-collar")]

list = loan_df[["age","job","marital","education"]]

print(list.sort_values(by="age"))
```
