````python
# Aggregate Functions
import pandas as pd

# Used to Summarize Data
# Perform Operations on Data
````

``` Whole DataFrame```

###### ```df.mean(numeric_only=True)``` -  Mean of Numeric Columns

###### ```df.sum(numeric_only=True)```  - Sum of Numeric Columns

###### ```df.min(numeric_only=True)```  - Minimum of Numeric Columns

###### ```df.max(numeric_only=True)```  - Maximum of Numeric Columns

###### ```df.count()```               - Count of Non-NA Cells for Each Column

````python
df = pd.read_csv("data_sets/population.csv")
print(df.mean(numeric_only=True))  # Mean of Numeric Columns
````

````python
df = pd.read_csv("data_sets/bank.csv")
````

````python
# Mean
print(df.mean(numeric_only=True))  # Mean of Numeric Columns
````

````python
# Sum
print(df.sum(numeric_only=True))  # Sum of Numeric Columns
````

``` Single Column```

###### ```df['column_name'].mean()``` -  Mean of a Specific Numeric Column

###### ```df['column_name'].sum()```  - Sum of a Specific Numeric Column

###### ```df['column_name'].min()```  - Minimum of a Specific Numeric Column

###### ```df['column_name'].max()```  - Maximum of a Specific Numeric Column

###### ```df['column_name'].count()```               - Count of Non-NA Cells for Each Column

````python
# Mean
print(df['age'].mean())  # Mean of Numeric Columns
````

````python
# Sum
print(df['balance'].sum())  # Sum of Numeric Columns
````

````python
# Minimum
print(df['age'].min())  # Minimum of Numeric Columns
````

````python
# Maximum
print(df['balance'].max())  # Maximum of Numeric Columns
````

````python
# Count
print(df['job'].count())
````

###### ```Note: Only a Numerical Columns will be Considered for Mean, Sum, Min, Max Functions. ```

``` Row-wise Aggregation```

```
df.mean(axis=1)  # Mean of Each Row
df.sum(axis=1)   # Sum of Each Row
df.min(axis=1)   # Minimum of Each Row
df.max(axis=1)   # Maximum of Each Row
```

````python
# Mean for row wise
df = pd.read_csv("data_sets/population.csv")

print(df.mean(axis=1,numeric_only=True))  # Mean of Each Row
````

````python
# Sum
print(df.sum(axis=1, numeric_only=True))  # Sum of Numeric Columns
````

````python
# Minimum
print(df.min(axis=1, numeric_only=True))  # Minimum of Numeric Columns
````

````python
# Maximum
print(df.max(axis=1, numeric_only=True))  # Maximum of Numeric Columns
````

``` Group By Aggregation```

```

````python
df = pd.read_csv("data_sets/insurance.csv")
````

````python
group = df.groupby('sex')   # Group by Gender
print(group['bmi'].mean())  # Mean Bmi for Each Gender Group
````

````python
# Sum
print(group['children'].sum())  # Sum of Children for Each Gender Group
````

````python
# Minimum
print(group['age'].min())  # Minimum Age for Each Gender
````

````python
# Maximum
print(group['age'].max())  # Maximum Age for Each Gender
````

````python
# Count
print(group['region'].count())  # Count of Each Region for Each
````
