##### ````Data cleaning ````
###### the process of fixing or removing incorrect

`````python
import pandas as pd
`````

`````python
df = pd.read_csv("data_sets/bank.csv")
`````

`````python
# 1. Drop irrelevant columns
df = df.drop(columns=["housing", "loan"])

print(df)
`````

`````python
# 2. Handle missing values

df = df.dropna(subset=["education"])
print(df.to_string())
`````

`````python
df = pd.read_csv("data_sets/bank.csv")
df = df.fillna({"education": "Non-Graduate"})
print(df.to_string())
`````

`````python
# 3. Fix Inconsistent values
df["education"] = df["education"].replace({"unknown": "Unknown", "secondary": "Secondary", "tertiary": "Tertiary", "primary": "Primary"})

print(df.to_string())
`````

`````python
# 4. Standardize data formats

df['job'] = df['job'].str.upper()
print(df.to_string())
`````

`````python
df['job'] = df['job'].str.lower()
print(df.to_string())
`````

`````python
# 5. Fix data types
df = pd.read_csv("data_sets/income.csv")
df['native-country'] = df['native-country'].astype('bool')
print(df.to_string())
`````

`````python
# 6. Remove duplicates

df = pd.read_csv("data_sets/bank.csv")
df = df.drop_duplicates()
print(df.to_string())
`````
