`````python
import matplotlib.pyplot as plt
import pandas as pd
`````

##### ````data_sets\bank.csv````

`````python
df = pd.read_csv("data_sets/bank.csv")  
`````

`````python
plt.pie(df['marital'].value_counts(),
        labels=df['marital'].value_counts().index,
        autopct='%1.1f%%',
        startangle=90,
        colors=['#ff9999','#66b3ff','#99ff99'],
        shadow=True
        )
plt.title('Marital Status Distribution')
plt.legend(title='Marital Status')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
`````

`````python
plt.bar(df['education'].value_counts().index,
        df['education'].value_counts(),
        color=['#ff9999','#66b3ff','#99ff99'],
        edgecolor='black'
        )
plt.xlabel('Education Level')
plt.ylabel('Count')
plt.title('Education Level Distribution')
plt.xticks(rotation=45)
plt.show()
`````

`````python
plt.hist(df['age'], color='#66b3ff', edgecolor='black')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Frequency')
plt.show()
`````

##### ```data_sets\insurance.csv```

`````python
df = pd.read_csv("data_sets/insurance.csv")
`````

`````python
plt.pie(df['region'].value_counts(),
        labels=df['region'].value_counts().index,
        autopct='%1.1f%%',
        startangle=90,
        colors=['#ff9999','#66b3ff','#99ff99', '#ffcc99'],
        shadow=True
        )
plt.title('Region Distribution')
plt.legend(title='Region')
plt.axis('equal')  # Equal aspect ratio ensures that pie is drawn as a circle.
plt.show()
`````

`````python
plt.bar(df['smoker'].value_counts().index, df['smoker'].value_counts(), color=["#99fff8","#ab72d1"])
plt.title('Smoker Distribution')
plt.xlabel('Smoker Status')
plt.ylabel('Count')
plt.show()
`````

`````python
plt.bar(df['age'].value_counts().index, 
        df['age'].value_counts(), 
        color=["#ff9999","#66b3ff","#99ff99", "#ffcc99", "#ffb3e6", "#c2f0c2", "#ff6666", "#c2d6d6", "#ffcc99", "#99ff99"])
plt.title('Age Distribution')
plt.xlabel('Age')
plt.xticks(rotation=45)
plt.ylabel('Count')
plt.show()
`````
