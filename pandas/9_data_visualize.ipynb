#### ````Marks Details````

`````python
import matplotlib.pyplot as plt
import pandas as pd
`````

`````python
df = pd.read_csv("student_marks.csv")
`````

`````python
print(df["IA_1"].max())
print(df["IA_1"].min()) 
`````

`````python
print(df["Final"].max())
print(df["Final"].min()) 
`````

`````python
print(df["Student_Name"][df["IA_1"] == df["IA_1"].max()])
`````

`````python
print(df["Student_Name"][df["IA_1"] == df["IA_1"].min()])
`````

`````python
plt.bar(df["Student_Name"][df["IA_1"] == df["IA_1"].max()], df["IA_1"][df["IA_1"] == df["IA_1"].max()])
plt.title('IA_1 Marks of Students')
plt.xlabel('Student Name')
plt.xticks(rotation=90)
plt.ylabel('IA_1 Marks')
plt.show()
`````

`````python
print(df["Student_Name"][df["Final"] == df["Final"].max()])
`````

`````python
print(df["Student_Name"][df["Final"] == df["Final"].min()])
`````

`````python
plt.pie(df["Assign_marks"].value_counts(),
        labels=df["Assign_marks"].value_counts().index, autopct='%1.1f%%')
plt.title('Assignment Marks Distribution')
plt.axis('equal')
plt.show()
`````

`````python
sorted_final = df["Final"].value_counts().sort_index()
colors = ['#ff9999','#66b3ff','#99ff99','#ffcc99','#c2c2f0','#ffb3e6','#c2f0c2','#ff6666','#66ffe0','#ffb366', '#c2c2f0','#ffb3e6','#c2f0c2','#ff6666','#66ffe0','#ffb366']
`````

`````python
plt.pie(sorted_final,
        labels=sorted_final.index,
        startangle=90,
        colors=colors,
        shadow=True,
        autopct='%1.1f%%')
plt.title('Final Marks Distribution')
plt.axis('equal')
plt.show()
`````
