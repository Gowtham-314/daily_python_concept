```DataFrames```

````python
# DataFrame = A tabular data structure with rows AND columns. (2 Dimensional) 
# Similar to an Excel spreadsheet
````

````python
import pandas as pd
````

````python
data = {"Name": ["Peter", "Steve", "Lois", "Mige"],
        "Age": [28, 24, 35, 32]
        }

df = pd.DataFrame(data, index=["Emp1", "Emp2", "Emp3", "Emp4"])
print(df)
````

````python
print(df.loc["Emp2"])  # Accessing a row by index label
````

````python
print(df.iloc[2])  # Accessing a row by index position
````

````python
# Add a New Column
df["Department"] = ["HR", "IT", "Finance", "Marketing"]
print(df)
````

````python
# Add a New Row 1
df.loc["Emp5"] = ["Brian", 30, "Sales"]
print(df)
````

````python
# Add a New Row 2

new_row = pd.DataFrame([{"Name": "Chris",
                        "Age": 29,
                        "Department": "Operations"  
                        }],
                       index=["Emp6"])
df = pd.concat([df, new_row])
print(df)
````

````python
# Add a New Rows

new_row = pd.DataFrame([{"Name": "Quagmire","Age": 29,"Department": "Graphics"},
                       {"Name": "Joe","Age": 31,"Department": "Logistics"}], index=["Emp7", "Emp8"])
df = pd.concat([df, new_row])
print(df)
````

````python
print(df[df['Age']>30])
````

````python
print(df.sort_values(by='Age'))  # Sorting by Age
````

````python
print(df.sort_index())  # Sorting by Index
````

````python
print(df.sort_values(by='Name'))  # Sorting by Name
````
