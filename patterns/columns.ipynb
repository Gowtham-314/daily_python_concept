``` Columns = Inner loop ```

````python
n = 5
````

````python
# Right Angle Triangle

for i in range(n):
    print(" "*(n-i)+" *"*(i+1))
````

````python
# Triangle

for i in range(n):
    print(" "*(i)+" *"*(n-i))
````

````python
# Triangle

for i in range(1,n+1):
    print(" "*(2*(n-i))+"*"*i)
````

````python
# Diamond

for i in range(n):
    print(" "*(n-i)+"*"*(2*i-1))

for i in range(n):
    print(" "*(i)+"*"*(2*(n-i)-1))
  
````

````python
n=5
for i in range(n):
    if i==0 or i==n-1: 
        print("* "*n)
    else:
        print("* "+" "*(n)+" *")
````
