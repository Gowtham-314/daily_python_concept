```Rows = Outer loop```

````python
# Square

n = int(input("Enter the number: "))
for i in range(n):
        print("* "*n)
````

````python
# Rectangle

n = int(input("Enter the number: "))
for i in range(n):
        print("* "*(n+1))
````

````python
# Right Angle Triangle

n = int(input("Enter the number: "))
for i in range(n):
    print("* "*i)
````

````python
# Right Angle Triangle

n = int(input("Enter the number: "))
for i in range(n):
    print("*"*int(i/n))
````

````python
# Right Angle Triangle

n = int(input("Enter the number: "))
for i in range(n):
    print("* "*(n-i))
````
