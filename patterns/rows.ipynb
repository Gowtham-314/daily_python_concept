```Rows = Outer loop```

```Square / Rectangle --->  n stars in each row```<br>
```Traingle --->  stars = row number(i)```<br>
```Reverse Traingle --->  stars = n - row + 1```<br>


````python
# Square

n = int(input("Enter the number: "))
for i in range(n):
        print("* "*n)
````

````python
# Rectangle

n = int(input("Enter the number: "))
for i in range(n):
        print("* "*(n+1))
````

````python
n = int(input("Enter the number: "))
````

````python
# Right Angle Triangle

for i in range(n):
    print("* "*i)
````

````python
# Left Triangle

for i in range(n):
    print("* "*(n-i))
````

````python
# Triangle

for i in range(n):
    print(" "*(i)+"*"*(n-i-1))
````

````python
# Triangle

for i in range(n):
    print(" "*(n-i-1)+"*"*(i))
````

````python
# Triangle
for i in range(n):
    print(" "*(n-i)+" *"*(i+1))
````

````python
# Reverse Triangle

for i in range(n):
    print(" "*(i)+" *"*(n-i))
````
